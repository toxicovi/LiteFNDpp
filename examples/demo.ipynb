# LiteFNDpp Demo Notebook

This notebook demonstrates how to use the **LiteFNDpp** package for fake news detection.

It includes:
- Initializing the model
- Preprocessing and feature extraction
- Making predictions on sample texts
- Generating explanations using LIME

```python
# Import the LiteFNDpp model
from litefndpp.model import LiteFNDpp

# Initialize the model
model = LiteFNDpp()
print("LiteFNDpp initialized successfully!")
```

## Sample texts
Let's create a few sample news headlines to test prediction and explanation.

```python
sample_texts = [
    "Breaking news! Scientists discovered a shocking cure for disease.",
    "Local community garden opens new sustainable project.",
    "Experts claim this new diet guarantees 100% weight loss!"
]
```

## Make predictions
We can predict whether each text is **Real (0)** or **Fake (1)**.

```python
predictions = model.predict(sample_texts)
for text, pred in zip(sample_texts, predictions):
    label = 'Fake' if pred == 1 else 'Real'
    print(f"{label}: {text}")
```

## Explanation with LIME and cognitive features
Get a detailed explanation for one of the texts.

```python
text_to_explain = sample_texts[0]
explanation = model.explain(text_to_explain, num_features=5)

print("Prediction:", 'Fake' if explanation['prediction'] == 1 else 'Real')
print("Top features influencing prediction:")
for feature, weight in explanation['top_features']:
    print(f"{feature}: {weight:.3f}")

print("\nCognitive Scores:")
for category, score in explanation['cognitive_scores'].items():
    print(f"{category}: {score:.3f}")
```
